## MEASURES OF CENTRAL TENDENCY
There are only three for example:
    -mean
    -Median
    -Mode

```python
student_data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Frank"],
    "Science": [89, 98, 79, 79, 69, 56],
    "math": [90, 95, 80, 85, 70, 60],
    "Gender": ["Female", "Male", "Male", "Male", "Female", "Male"]
}
students = pd.DataFrame(student_data)
students
```

```python
students["math"].mean()
```

```python
students["math"].median()
```

```python
students["math"].mode()[0]
```
