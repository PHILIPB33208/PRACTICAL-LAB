(a) What is the sample space for:
1. 3 coins 
2. 4 coins 
3. 5 coins 

How many outcomes are there in each case?

From the sample space, we can define events. For example , the event "exactly 2 heads is a subset of the sample space 

```python
from itertools import product
```

```python
def generate_sample_space(n):
    return [''.join(x) for x in product('HT', repeat=n)]

for n in [3, 4, 5]:
    sample = generate_sample_space(n)
    print(f"SAMPLE SPACE FOR {n} COINS")
    print(sample)
    print(f"Total Outcomes =", len(sample))

    exactly_two_heads = [x for x in sample if x.count('H') == 2]

    print(f"Event: Exactly 2 Heads")

```

```python
if len(exactly_two_heads) > 0:
        print(exactly_two_heads)
        print("Number of Event Outcomes =", len(exactly_two_heads))
```

using outcomes by hand gets tiring quickily, itertools, product builds every combination for us.
what is the sample space for;
1. 3 coin
2. 4 coins
3. 3 dices?
how many outcomes are there in each case?

```python
from itertools import product
from fractions import Fraction

def prob(event, space):
    return Fraction(len(event), len(space))
coins3 = ['' .join(x) for x in product('HT', repeat=3)]
coins4 = [''.join(x) for x in product('HT', repeat=4)]
dice2 = list(product(range(1, 7), repeat=2))
```

```python
at_least_one = [x for x in coins3 if 'H' in x]
print("Direct:", prob(at_least_one, coins3))            

## Complement
not_E = [x for x in coins3 if 'H' not in x]              
print("Complement:", 1 - prob(not_E, coins3))            

## Extra probabilities
print("3 coins, exactly 2 heads:", prob([x for x in coins3 if x.count('H') == 2], coins3))  # 3/8
print("4 coins, exactly 2 heads:", prob([x for x in coins4 if x.count('H') == 2], coins4))  # 3/8
print("4 coins, at least 3 heads:", prob([x for x in coins4 if x.count('H') >= 3], coins4)) # 5/16
print("4 coins, all same:", prob([x for x in coins4 if len(set(x)) == 1], coins4))          # 1/8
print("2 dice, sum = 7:", prob([x for x in dice2 if sum(x) == 7], dice2))                   # 1/6
print("2 dice, doubles:", prob([x for x in dice2 if x[0] == x[1]], dice2))                  # 1/6
print("2 dice, sum >= 10:", prob([x for x in dice2 if sum(x) >= 10], dice2))                # 1/6
print("2 dice, at least one 6:", prob([x for x in dice2 if 6 in x], dice2))                 # 11/36
```
