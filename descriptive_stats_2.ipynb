1. ##female students number who scored 80

```python
import pandas as pd
student_data = pd.read_csv('studentsPerformance.csv')
```

```python
student_data[
    (student_data['gender'] == 'female') &
    (student_data['math score'] >= 80)
].shape[0]
```

2.mean

```python
student_data[
    student_data['test preparation course'] == 'completed'
]['reading score'].mean()
```

## 3. range of writing score

```python
student_data['writing score'].max() - student_data['writing score'].min()
```

## 4. Standard deviation of male students in reading score

```python
student_data[
    student_data['gender'] == 'male'
]['reading score'].std()
```

## 5. Create the total score

Add Mathematics + Reading + Writing

```python
student_data['total score'] = (
    student_data['math score'] +
    student_data['reading score'] +
    student_data['writing score']
)
```

5a) #### Average of the total score

```python
student_data['total score'].mean()
```

5b) ### Ninth percentile of total score

```python
student_data['total score'].quantile(0.09)
```

5c) ### Quantile range (IQR)

```python
Q1 = student_data['total score'].quantile(0.25)
Q3 = student_data['total score'].quantile(0.75)

Q3 - Q1
```
